# 3D Matérn Random Fields on the Norne Reservoir

This notebook walks through `examples/random_fields/matern_tpfa_norne_3d.jl`, which demonstrates how to:

1. Load the real-world **Norne** reservoir grid from a `.DATA` file via `GeoEnergyIO`.
2. Define a **3D Matérn SPDE prior** — a mesh-adapted Gaussian random field that respects the unstructured IJK topology of a reservoir.
3. Optionally apply **variance compensation** to correct boundary-induced marginal-variance bias.
4. Build the sparse **SPDE precision operator** with a **halo layer** of ghost cells to suppress boundary artefacts.
5. Draw a **sample** from the field and visualise it on three orthogonal slices.
6. Compute **diagnostic statistics** (realized variance and correlation-length ranges) to verify the prior.
7. Plot a **semivariogram** along each IJK axis to inspect spatial correlation structure, including the halo region.

All the heavy lifting is done by `JutulDarcy` functions:
`MaternSPDE3Prior`, `matern_sd_compensation!`, `matern_spde_operator`,
`matern_realized_variance`, and `matern_realized_axis_ranges`.

````julia
# ── Packages ──────────────────────────────────────────────────────────────────
using Jutul           # Core numerical framework (mesh, grids, linear algebra)
using JutulDarcy      # Porous-media extension: Matérn SPDE priors, operators, diagnostics
using Jutul: si_unit  # Physical unit helper: si_unit(:meter), etc.
using CairoMakie      # Static vector / raster plotting
using SparseArrays    # Needed for sparse precision matrices
using LinearAlgebra   # norm, dot, Symmetric, …
using Random          # RNG control for reproducible samples
using Statistics      # mean, std, quantile
using GeoEnergyIO     # I/O for Eclipse-format reservoir DATA files (Norne)
````

## 1 — Load the Norne Reservoir Grid

Norne is a real North Sea oil field. `GeoEnergyIO` ships an anonymised test version of the ECLIPSE input deck (`NORNE_NOHYST.DATA`). `parse_data_file` reads the complete deck; `reservoir_domain` extracts the **active** grid cells together with their geometry and properties.

> **Why TPFA?** The script title includes "tpfa" because the SPDE operator is assembled on the same two-point flux approximation (TPFA) connectivity graph used for the flow simulation. This means the random field naturally honours the mesh topology — inactive cells and pinch-outs are excluded automatically.

````julia
function load_norne_domain_3d()
    norne_dir = GeoEnergyIO.test_input_file_path("NORNE_NOHYST")
    data_pth  = joinpath(norne_dir, "NORNE_NOHYST.DATA")
    data      = parse_data_file(data_pth)
    return reservoir_domain(data)
end

domain = load_norne_domain_3d()
println("Number of active cells: ", number_of_cells(domain))
````

## 2 — Define the Matérn SPDE Prior

### Background: Matérn fields via the SPDE approach

A stationary Matérn covariance with range $\rho$ and marginal standard deviation $\sigma$ can be represented as the solution to the stochastic PDE

$$( \kappa^2 I - \Delta )^{\alpha/2} \, u = \mathcal{W}$$

where $\mathcal{W}$ is Gaussian white noise and $\kappa = \sqrt{8\nu}/\rho$ for smoothness $\nu$.

On a mesh the operator discretises into a **sparse precision matrix**

$$Q = D_\tau \, K \, C^{-1} K \, D_\tau$$

where $K$ is the finite-element stiffness matrix assembled on the TPFA graph, $C$ is the diagonal mass (cell-volume) matrix, and $D_\tau$ encodes spatially-varying precision scaling.

`MaternSPDE3Prior` stores the prior parameters but does **not yet** build $Q$. That happens in §4 when we call `matern_spde_operator`.

### Anisotropy

An optional axis-aligned anisotropy tensor

$$H = \operatorname{diag}\!\bigl(e^{a}, e^{b}, e^{-(a+b)}\bigr), \quad \det H = 1$$

stretches or compresses the effective correlation lengths per axis:

$$\rho_I = \rho_0 e^{a/2}, \quad \rho_J = \rho_0 e^{b/2}, \quad \rho_K = \rho_0 e^{-(a+b)/2}.$$

In isotropic mode (`aniso_a = aniso_b = 0`) all three ranges equal $\rho_0$.

````julia
function build_prior_spec_3d(domain; aniso_a = 0.0, aniso_b = 0.0)
    nc  = number_of_cells(domain)
    ρ0  = 1500.0 * si_unit(:meter)   # nominal isotropic correlation length
    σ0  = 1.0                          # marginal standard deviation

    has_aniso = aniso_a != 0.0 || aniso_b != 0.0
    if has_aniso
        # Basis matrices: one column per basis function; here just a constant field
        Ba    = ones(Float64, nc, 1)
        Bb    = ones(Float64, nc, 1)
        prior = MaternSPDE3Prior(nc; ρ0 = ρ0, σ0 = σ0,
                                 anisotropy_a_basis = Ba, anisotropy_b_basis = Bb)
        return prior, (anisotropy_a = [Float64(aniso_a)], anisotropy_b = [Float64(aniso_b)])
    else
        prior = MaternSPDE3Prior(nc; ρ0 = ρ0, σ0 = σ0)
        return prior, NamedTuple()
    end
end

# Isotropic prior (no anisotropy)
prior, θ = build_prior_spec_3d(domain)
println("Prior: ρ0 = ", prior.ρ0, " m,  σ0 = ", prior.σ0)
````

## 3 — Lattice Anchors for Diagnostics

Diagnostics (realized variance, realized correlation ranges) require evaluating the marginal properties of the field at a finite set of representative cells. Rather than picking cells randomly, we tile a regular $n_x \times n_y \times n_z$ lattice over the domain bounding box and snap each lattice node to the nearest active cell centroid.

````julia
function lattice_anchors_3d(domain; nx = 3, ny = 3, nz = 2)
    cc = domain[:cell_centroids]
    xs = range(minimum(cc[1, :]), maximum(cc[1, :]), length = nx)
    ys = range(minimum(cc[2, :]), maximum(cc[2, :]), length = ny)
    zs = range(minimum(cc[3, :]), maximum(cc[3, :]), length = nz)
    anchors = Int[]
    for x in xs, y in ys, z in zs
        d2 = [(cc[1, i] - x)^2 + (cc[2, i] - y)^2 + (cc[3, i] - z)^2 for i in axes(cc, 2)]
        push!(anchors, argmin(d2))
    end
    return unique(anchors)
end

anchors = lattice_anchors_3d(domain; nx = 3, ny = 3, nz = 2)
println("Number of diagnostic anchor cells: ", length(anchors))
````

## 4 — Variance Compensation (optional)

Near the boundary of the domain the SPDE operator has fewer neighbours, which reduces the effective integration volume and causes the **marginal variance to be lower** than the target $\sigma_0^2$.

`matern_sd_compensation!` corrects for this by:

1. Evaluating the **realized diagonal** of $Q^{-1}$ at the anchor cells using an efficient $K$-factorisation approach (numerically more stable than factorising $Q$ directly on large grids).
2. Fitting spatially-varying scaling coefficients so that the corrected marginal variance matches $\sigma_0^2$ at the anchors.
3. Storing the compensation coefficients in the `prior` object in-place.

The `halo_spec` adds ghost cells beyond the active domain boundary to further soften boundary effects before compensation even kicks in. `layers` sets how many layers of halo cells to add; `growth` is the geometric factor by which each successive halo layer expands in size.

````julia
# Halo spec: 3 layers of ghost cells beyond active boundary, each 1.5× bigger
halo_spec = MaternHaloSpec(layers = 3, growth = 1.5)

# Set compensate = true to apply boundary-variance correction
compensate = false   # ← toggle to true to enable

if compensate
    matern_sd_compensation!(prior, domain, θ; mode = :mean, anchors = anchors, halo = halo_spec)
    println("Compensation coefficients applied to prior.")
else
    println("Running without compensation.")
end
````

## 5 — Assemble the SPDE Operator

`matern_spde_operator` assembles the full sparse precision matrix $Q$ (including halo cells), the Cholesky-like factorisation needed for sampling, and all metadata about the halo topology.

The returned `op` struct contains:
- `op.Q` — the symmetric sparse precision matrix (reservoir + halo cells)
- `op.K` — the stiffness matrix factor (used for efficient linear solves)
- `op.C` — the diagonal mass matrix
- `op.fields.τ` — per-cell precision scaling from $D_\tau$
- `op.halo_idx` — indices of halo cells within the extended system
- `op.halo_meta` — connectivity metadata for the halo (face-to-chain mapping, extended centroids)

````julia
op = matern_spde_operator(domain, prior, θ; compensated = compensate, halo = halo_spec)

nc_total = size(op.Q, 1)
println("Reservoir cells : ", number_of_cells(domain))
println("Halo cells      : ", length(op.halo_idx))
println("Total SPDE size : ", nc_total)
````

## 6 — Draw a Sample

Sampling from $\mathcal{N}(0, Q^{-1})$ uses the factored precision:

$$x = Q^{-1/2} \xi = D_\tau^{-1} K^{-1} \sqrt{C}\, \xi, \quad \xi \sim \mathcal{N}(0, I)$$

This is a single sparse triangular solve — very fast even for 100 k+ unknowns.

After sampling we mask out the halo cells with `NaN` so they don't contaminate the slice plots.

````julia
function sample_matern(op; rng = Random.default_rng())
    n       = size(op.K, 1)
    ξ       = randn(rng, n)
    sqrtC_ξ = sqrt.(op.C.diag) .* ξ
    y       = op.K \ sqrtC_ξ
    τ_vec   = op.fields.τ
    return y ./ τ_vec
end

rng = MersenneTwister(42)    # fix seed for reproducibility
x   = sample_matern(op; rng = rng)

# Mask halo cells
x_masked              = copy(x)
x_masked[op.halo_idx] .= NaN

nc          = number_of_cells(domain)
finite_vals = x_masked[1:nc][isfinite.(x_masked[1:nc])]
println("Sample statistics (reservoir only):")
println("  mean = ", round(mean(finite_vals), digits = 4))
println("  std  = ", round(std(finite_vals),  digits = 4))
````

## 7 — Visualise Sample Slices

`plot_sample_slices` extracts three orthogonal mid-plane slices (I, J, K) through the active cells and scatter-plots the sample values coloured by magnitude. Robust colour scaling at the 98th-percentile absolute value avoids saturation from outliers.

````julia
# ── Helper: robust colour scale ────────────────────────────────────────────────
function _robust_abs_scale(x; robust_quantile = 0.98)
    finite_values = x[isfinite.(x)]
    isempty(finite_values) && return 1.0
    abs_values = abs.(finite_values)
    x_abs = robust_quantile < 1 ? quantile(abs_values, robust_quantile) : maximum(abs_values)
    return max(x_abs, eps(Float64))
end

# ── Helper: IJK mid-plane indices ─────────────────────────────────────────────
function _slice_indices(mesh)
    ijk   = map(cell -> cell_ijk(mesh, cell), 1:number_of_cells(mesh))
    is    = sort(unique(first.(ijk)))
    js    = sort(unique(map(x -> x[2], ijk)))
    ks    = sort(unique(last.(ijk)))
    i_mid = is[cld(length(is), 2)]
    j_mid = js[cld(length(js), 2)]
    k_mid = ks[cld(length(ks), 2)]
    return (ijk = ijk, i_mid = i_mid, j_mid = j_mid, k_mid = k_mid)
end

function _cells_on_slice(mesh, ijk_cache, axis::Symbol, index::Int)
    cells = Int[]
    for (cell, ijk) in enumerate(ijk_cache)
        if     axis == :i && ijk[1] == index; push!(cells, cell)
        elseif axis == :j && ijk[2] == index; push!(cells, cell)
        elseif axis == :k && ijk[3] == index; push!(cells, cell)
        end
    end
    return cells
end

function _slice_axis(figpos, title; orientation::Symbol)
    if orientation == :top
        return Axis(figpos, title = title, xlabel = "x [m]", ylabel = "y [m]", aspect = DataAspect())
    elseif orientation == :xz
        return Axis(figpos, title = title, xlabel = "x [m]", ylabel = "z [m]", yreversed = true, aspect = DataAspect())
    elseif orientation == :yz
        return Axis(figpos, title = title, xlabel = "y [m]", ylabel = "z [m]", yreversed = true, aspect = DataAspect())
    end
end
````

````julia
function plot_sample_slices(domain, x;
        output_path     = "sample_slices.png",
        title           = "3D Matérn sample",
        colormap        = :RdBu,
        figure_size     = (1400, 900),
        robust_quantile = 0.98)

    mesh   = physical_representation(domain)
    nc     = number_of_cells(domain)
    values = length(x) > nc ? x[1:nc] : copy(x)
    cc     = Matrix{Float64}(domain[:cell_centroids])
    cache  = _slice_indices(mesh)

    cells_k   = _cells_on_slice(mesh, cache.ijk, :k, cache.k_mid)
    cells_j   = _cells_on_slice(mesh, cache.ijk, :j, cache.j_mid)
    cells_i   = _cells_on_slice(mesh, cache.ijk, :i, cache.i_mid)
    color_abs = _robust_abs_scale(values; robust_quantile = robust_quantile)

    fig = Figure(size = figure_size)
    axk = _slice_axis(fig[1, 1], "$title: K = $(cache.k_mid)", orientation = :top)
    axj = _slice_axis(fig[1, 2], "$title: J = $(cache.j_mid)", orientation = :xz)
    axi = _slice_axis(fig[2, 1], "$title: I = $(cache.i_mid)", orientation = :yz)

    plt = nothing
    for (ax, cells, dims) in ((axk, cells_k, (1, 2)), (axj, cells_j, (1, 3)), (axi, cells_i, (2, 3)))
        plt = scatter!(ax, cc[dims[1], cells], cc[dims[2], cells],
                       color      = values[cells],
                       colormap   = colormap,
                       colorrange = (-color_abs, color_abs),
                       markersize = 9)
    end
    Colorbar(fig[1:2, 3], plt)
    save(output_path, fig)
    return (fig = fig, output_path = output_path)
end

sample_plot = plot_sample_slices(domain, x_masked;
    output_path = "matern_sample_slices.png",
    title       = "Norne 3D Matérn sample")
println("Saved → ", sample_plot.output_path)
sample_plot.fig
````

## 8 — Diagnostic Plots: Variance and Axis Ranges

These diagnostics answer two questions:

| Diagnostic | What it checks |
|---|---|
| **Realized variance** | Is the marginal $\operatorname{Var}(u_i) \approx \sigma_0^2$ at each anchor? |
| **Realized axis ranges** | Is the correlation length in each IJK direction close to the theoretical value? |

Both quantities are computed by evaluating a small subset of $Q^{-1}$ — the covariance matrix — using efficient solves through the $K$ factor. (Direct Cholesky on $Q$ is fragile for large grids because $Q = D_\tau K C^{-1} K D_\tau$ is not numerically SPD when the diagonal of $Q$ spans many orders of magnitude, as on the full Norne grid.)

`corr_level = 0.2` means the range is reported as the lag at which correlation drops to 0.2.

````julia
function plot_prior_diagnostics_3d(variance_diag, axis_diag;
        output_path = "matern_diagnostics.png",
        title       = "3D Matérn diagnostics",
        figure_size = (1200, 450))

    n   = length(variance_diag.anchors)
    ix  = 1:n
    fig = Figure(size = figure_size)

    # ── Variance panel ────────────────────────────────────────────────────────
    ax1 = Axis(fig[1, 1], title = "$title: Variance",
               xlabel = "Anchor index", ylabel = "Variance")
    lines!(ax1, ix, variance_diag.target,   color = :black,      linestyle = :dash, label = "Target")
    lines!(ax1, ix, variance_diag.realized, color = :dodgerblue, linewidth = 2,     label = "Realized")
    scatter!(ax1, ix, variance_diag.realized, color = :dodgerblue, markersize = 8)
    axislegend(ax1, position = :rb)

    # ── Range panel ───────────────────────────────────────────────────────────
    ax2 = Axis(fig[1, 2], title = "$title: Axis ranges",
               xlabel = "Anchor index", ylabel = "Range [m]")
    for (color, dash_label, real_label, tgt, rlz) in (
            (:firebrick,   "Target x", "Realized x", axis_diag.target_x, axis_diag.realized_x),
            (:forestgreen, "Target y", "Realized y", axis_diag.target_y, axis_diag.realized_y),
            (:darkorange,  "Target z", "Realized z", axis_diag.target_z, axis_diag.realized_z))
        lines!(ax2,  ix, tgt, color = color, linestyle = :dash, label = dash_label)
        lines!(ax2,  ix, rlz, color = color, linewidth = 2,     label = real_label)
        scatter!(ax2, ix, rlz, color = color, markersize = 8)
    end
    axislegend(ax2, position = :rb)

    save(output_path, fig)
    return (fig = fig, output_path = output_path)
end

variance_diag = matern_realized_variance(domain, prior, θ;
    anchors = anchors, compensated = compensate, halo = halo_spec)

axis_diag = matern_realized_axis_ranges(domain, prior, θ;
    anchors = anchors, compensated = compensate, halo = halo_spec, corr_level = 0.2)

diag_plot = plot_prior_diagnostics_3d(variance_diag, axis_diag;
    output_path = "notebook/matern_diagnostics.png")
println("Saved → ", diag_plot.output_path)
diag_plot.fig
````

## 9 — Semivariogram Along IJK Axes

The **semivariogram** $\gamma(h)$ measures how quickly spatial correlation decays with lag distance $h$:

$$\gamma(h) = \tfrac{1}{2}\operatorname{Var}\!\bigl(u(x) - u(x+h)\bigr) = \sigma^2\bigl(1 - C(h)/\sigma^2\bigr)$$

where $C(h)$ is the covariance at lag $h$.

`variogram_line_3d` walks from the midpoint of the grid toward the boundary along one IJK axis, collecting active cells (and the halo chain beyond the last active cell), then computes the semivariogram from the dense covariance sub-matrix $\Sigma = Q^{-1}[\text{idx}, \text{idx}]$.

The plot shades the **halo region** in grey and marks:
- The first **inactive cell** (gap in the IJK walk due to a pinch-out or pillar deactivation) with a dash-dot vertical line.
- The **halo start** (where ghost cells begin) with a dotted vertical line.

These markers help you see how the halo extends spatial correlation smoothly through the boundary rather than cutting it abruptly.

````julia
function covariance_subset_from_precision(Q::Symmetric; idx::AbstractVector{<:Integer})
    nc = size(Q, 1)
    m  = length(idx)
    E  = zeros(Float64, nc, m)
    for (k, j) in enumerate(idx); E[j, k] = 1.0; end
    F  = JutulDarcy._factor_precision_matrix(Q)
    U  = F \ E
    return Symmetric(U[idx, :])
end

function variogram_line_3d(domain, op; direction::Symbol = :j)
    mesh = physical_representation(domain)
    nc   = number_of_cells(domain)
    cc   = Matrix{Float64}(domain[:cell_centroids])

    ijk  = map(cell -> cell_ijk(mesh, cell), 1:nc)
    is   = sort(unique(first.(ijk)))
    js   = sort(unique(map(x -> x[2], ijk)))
    ks   = sort(unique(last.(ijk)))
    nx, ny, nz = grid_dims_ijk(mesh)

    i_mid = is[cld(length(is), 2)]
    j_mid = js[cld(length(js), 2)]
    k_mid = ks[cld(length(ks), 2)]
    active_lookup = Dict(ijk[c] => c for c in 1:nc)

    if direction == :i
        line_positions = collect(i_mid:nx);  make_key = pos -> (pos, j_mid, k_mid)
    elseif direction == :j
        line_positions = collect(j_mid:ny);  make_key = pos -> (i_mid, pos, k_mid)
    else
        line_positions = collect(k_mid:nz);  make_key = pos -> (i_mid, j_mid, pos)
    end

    idx_reservoir = [active_lookup[make_key(pos)]
                     for pos in line_positions if haskey(active_lookup, make_key(pos))]
    isempty(idx_reservoir) && error("No active cells along $direction.")

    ref_cell    = first(idx_reservoir)
    ref_point   = cc[:, ref_cell]
    last_active = last(idx_reservoir)

    # Halo chain from the last active cell along the walk direction
    halo_chain = Int[]
    halo_meta  = op.halo_meta
    if !isnothing(halo_meta)
        bn  = Int.(vec(domain[:boundary_neighbors]))
        bc  = Matrix{Float64}(domain[:boundary_centroids])
        face_to_chain = halo_meta.face_to_chain
        dir_vec = zeros(3)
        if direction == :i; dir_vec[1] = 1.0
        elseif direction == :j; dir_vec[2] = 1.0
        else  dir_vec[3] = 1.0; end

        best_face, best_dot = 0, -Inf
        for f in eachindex(bn)
            bn[f] == last_active           || continue
            haskey(face_to_chain, f)        || continue
            any(!isfinite, bc[:, f])        && continue
            outward = bc[:, f] .- cc[:, last_active]
            d = dot(outward, dir_vec)
            if d > best_dot; best_dot = d; best_face = f; end
        end
        best_face > 0 && (halo_chain = face_to_chain[best_face])
    end

    idx_all = vcat(idx_reservoir, halo_chain)
    Σ       = covariance_subset_from_precision(op.Q; idx = idx_all)
    ref_var = Σ[1, 1]

    lags, γ_values, first_inactive_lag = Float64[], Float64[], NaN
    for pos in line_positions
        key = make_key(pos)
        if haskey(active_lookup, key)
            cell     = active_lookup[key]
            lag      = norm(cc[:, cell] - ref_point)
            si       = findfirst(==(cell), idx_all)
            cov_val  = Σ[1, si]
            cell_var = Σ[si, si]
            push!(lags, lag)
            push!(γ_values, 0.5 * (ref_var + cell_var - 2 * cov_val))
        else
            isnan(first_inactive_lag) && !isempty(lags) && (first_inactive_lag = lags[end])
            push!(lags, NaN); push!(γ_values, NaN)
        end
    end

    reservoir_end_lag = norm(cc[:, last_active] - ref_point)
    halo_start_lag    = NaN
    if !isempty(halo_chain)
        ext_centroids = halo_meta.ext_centroids
        for (k, cell) in enumerate(halo_chain)
            si      = findfirst(==(cell), idx_all)
            cpt     = ext_centroids[:, cell]
            lag     = norm(cpt - ref_point)
            cov_val = Σ[1, si]; cell_var = Σ[si, si]
            k == 1 && (halo_start_lag = lag)
            push!(lags, lag)
            push!(γ_values, 0.5 * (ref_var + cell_var - 2 * cov_val))
        end
    end

    return (lags = lags, γ = γ_values,
            reservoir_end_lag  = reservoir_end_lag,
            halo_start_lag     = halo_start_lag,
            first_inactive_lag = first_inactive_lag,
            direction          = direction)
end

println("Computing variograms along I, J, K axes …")
vario_i = variogram_line_3d(domain, op; direction = :i)
vario_j = variogram_line_3d(domain, op; direction = :j)
vario_k = variogram_line_3d(domain, op; direction = :k)
println("Done.")
````

````julia
function plot_variogram_3d(vario_i, vario_j, vario_k;
        output_path = "matern_variogram.png",
        figure_size = (1500, 450))

    fig    = Figure(size = figure_size)
    colors = (:firebrick, :forestgreen, :darkorange)
    labels = ("I-direction", "J-direction", "K-direction")

    for (col, vario, color, label) in zip(1:3, (vario_i, vario_j, vario_k), colors, labels)
        ax = Axis(fig[1, col], title = "Variogram: $label",
                  xlabel = "Lag distance [m]", ylabel = "Semivariogram")
        lags, γ = vario.lags, vario.γ
        valid   = isfinite.(lags) .& isfinite.(γ)

        if isfinite(vario.halo_start_lag) && any(valid)
            max_lag = maximum(lags[valid])
            max_lag > vario.halo_start_lag &&
                vspan!(ax, vario.halo_start_lag, max_lag, color = (:gray85, 0.4))
        end
        isfinite(vario.halo_start_lag) &&
            vlines!(ax, [vario.halo_start_lag],    color = :gray25, linestyle = :dot,     linewidth = 2, label = "Halo start")
        isfinite(vario.first_inactive_lag) &&
            vlines!(ax, [vario.first_inactive_lag], color = :purple, linestyle = :dashdot, linewidth = 2, label = "First inactive")

        scatter!(ax, lags[valid], γ[valid], color = color, markersize = 7)
        for k in 1:(length(lags) - 1)
            if isfinite(lags[k]) && isfinite(lags[k+1]) && isfinite(γ[k]) && isfinite(γ[k+1])
                lines!(ax, [lags[k], lags[k+1]], [γ[k], γ[k+1]], color = color, linewidth = 2)
            end
        end
        axislegend(ax, position = :rb)
    end

    save(output_path, fig)
    println("Saved variogram plot to ", output_path)
    return (fig = fig, output_path = output_path)
end

vario_plot = plot_variogram_3d(vario_i, vario_j, vario_k;
    output_path = "matern_variogram.png")
vario_plot.fig
````

## 10 — Putting It All Together: the `main` Function

The example script wraps all of the above in a single `main` function callable from the REPL or from the command line with positional arguments:

```
julia --project=./ examples/random_fields/matern_tpfa_norne_3d.jl <compensate> <halo_layers> <halo_growth> <aniso_a> <aniso_b>
```

| Argument | Default | Meaning |
|---|---|---|
| `compensate` | `false` | No variance compensation |
| `halo_layers` | `3` | Three ghost-cell layers |
| `halo_growth` | `1.5` | Each layer 1.5× the previous |
| `aniso_a` | `0.0` | No x/y anisotropy |
| `aniso_b` | `0.0` | No y/z anisotropy |

The cell below demonstrates running the full pipeline programmatically with configurable parameters — toggle `compensate`, or set non-zero values for `aniso_a` / `aniso_b` to explore anisotropic priors.

````julia
# ── Configuration ─────────────────────────────────────────────────────────────
compensate  = false   # set true to enable boundary-variance correction
halo_layers = 3
halo_growth = 1.5
aniso_a     = 0.0     # e.g. 1.0 to elongate along I-axis
aniso_b     = 0.0     # e.g. -0.5 to compress along J-axis

# ── Pipeline ──────────────────────────────────────────────────────────────────
domain_run  = load_norne_domain_3d()
prior_run, θ_run = build_prior_spec_3d(domain_run; aniso_a = aniso_a, aniso_b = aniso_b)
halo_run    = MaternHaloSpec(layers = halo_layers, growth = halo_growth)
anchors_run = lattice_anchors_3d(domain_run; nx = 3, ny = 3, nz = 2)

if compensate
    matern_sd_compensation!(prior_run, domain_run, θ_run;
        mode = :mean, anchors = anchors_run, halo = halo_run)
end

op_run  = matern_spde_operator(domain_run, prior_run, θ_run;
    compensated = compensate, halo = halo_run)
x_run   = sample_matern(op_run; rng = MersenneTwister(1))
x_run[op_run.halo_idx] .= NaN

fig_run = plot_sample_slices(domain_run, x_run;
    output_path = "main_sample.png",
    title       = "Norne 3D — compensate=$compensate, a=$aniso_a, b=$aniso_b")

println("Norne 3D: nc=$(number_of_cells(domain_run)),  ",
        "total SPDE size=$(size(op_run.Q, 1))")
println("Compensation: $compensate")
if aniso_a != 0.0 || aniso_b != 0.0
    ρ0 = prior_run.ρ0
    println("Effective ranges: ρ_I=$(ρ0*exp(aniso_a/2)),  ",
            "ρ_J=$(ρ0*exp(aniso_b/2)),  ",
            "ρ_K=$(ρ0*exp(-(aniso_a+aniso_b)/2))")
end
fig_run.fig
````
